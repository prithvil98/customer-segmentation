```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
rfm= pd.read_csv("../data/processed/rfm_segments.csv")
```

```python
rfm.head()
```

```python
rfm.shape
```

```python
rfm.describe()
```

```python
#COunt Plot
plt.figure(figsize=(4, 4))
sns.countplot(data=rfm, x="Segment", order=["Champions", "Loyal", "At Risk", "Lost"])
plt.title("Number of Customers per Segment")
plt.ylabel("Customer Count")
plt.show()
```

```python
#Distribution Charts
fig, axes = plt.subplots(1, 3, figsize=(16, 4))

sns.histplot(rfm["Recency"], bins=30, ax=axes[0])
axes[0].set_title("Recency Distribution")

sns.histplot(rfm["Frequency"], bins=30, ax=axes[1])
axes[1].set_title("Frequency Distribution")

sns.histplot(rfm["Monetary"], bins=30, ax=axes[2])
axes[2].set_title("Monetary Distribution")

plt.tight_layout()
plt.show()
```
